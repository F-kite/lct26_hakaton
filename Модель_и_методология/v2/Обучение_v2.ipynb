# Обучение модели v2

Здесь показан простой принцип: загрузить числовые признаки, обучить классификатор и проверить качество. Для поиска по смыслу используется готовый локальный E5 в `signals.py`.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

data = pd.read_csv('Данные/Признаки.csv')
labels = pd.read_csv('Данные/Разметка.csv')[['id','label','partition']]
data = data.merge(labels, on='id')
X = data.filter(regex=r'^x')
y = data['label']
train = data['partition'].isin(['train','validation'])
test = data['partition'].eq('new_test')
print(X.shape, y.value_counts().to_dict())
```

```python
model = LogisticRegression(C=100, class_weight='balanced', max_iter=3000)
model.fit(X[train], y[train])
pred = model.predict(X[test])
print(classification_report(y[test], pred, zero_division=0))
```
